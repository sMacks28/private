```python
from sympy import symbols, Matrix, nsimplify, simplify
```

```python
D, U_p, U_pp, phi_1x, phi_1y, phi_2x, phi_2y, delta, theta, k, p = symbols('D U_p U_pp phi_1x phi_1y phi_2x phi_2y delta theta k p')
```

```python
matrix = Matrix([[(-1/2*D**2 + 2*U_p + 4*phi_1x**2*U_pp)*delta + (k**2 - p**2)*theta, 4*phi_2x*phi_1y*U_pp],
          [4*phi_1x*phi_2y*U_pp, (-1/2*D**2 + 2*U_p + 4*phi_2x**2*U_pp)*delta + (k**2 - p**2)*theta]])
```

```python
matrix_inv = nsimplify(simplify(matrix.inv()))
```
