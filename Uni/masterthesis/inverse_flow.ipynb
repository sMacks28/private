```python
from sympy import symbols, Matrix, nsimplify, simplify
```

```python
D, U_p, U_pp, phi_1, phi_2, k, p, R_k = symbols('D U_p U_pp phi_1 phi_2 k p R_k')
```

```python
a = 2*U_p + 4*phi_1**2*U_pp + k**2
b = -4*phi_1*phi_2*U_pp
c = -4*phi_1*phi_2*U_pp
d = 2*U_p + 4*phi_2**2*U_pp + k**2

```

```python
simplify(a*d - b*c)
```

```python
nsimplify(1/2 * (a + d))
```
